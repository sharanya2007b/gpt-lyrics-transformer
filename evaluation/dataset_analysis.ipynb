```python
from collections import Counter

def repetition_ratio(text):
    words = text.split()
    counts = Counter(words)

    repeated = sum(v for v in counts.values() if v > 1)

    return repeated / len(words)

print("Repetition Ratio:", repetition_ratio(text))
```
